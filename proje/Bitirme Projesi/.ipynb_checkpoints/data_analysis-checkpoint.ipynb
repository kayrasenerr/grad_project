```python
import os

import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
sample_dir = "samples"
sesdyn_data_path = os.path.join("BehaviorDatasets", "SESDYN_Basic_A",
                                sample_dir, "Basic_A_Complete.xlsx")
sesdyn_df = pd.read_excel(sesdyn_data_path)
sesdyn_df = sesdyn_df.rename(columns={"Final": 100})
sesdyn_df
```

```python
sesdyn_df['Batch'].unique()
```

```python
for type in sesdyn_df["Label"].unique():
    sesdyn_type_df = sesdyn_df[sesdyn_df["Label"]==type]
    sesdyn_type_df = sesdyn_type_df.drop(
        ["Instance No", "Batch", "Label", "ID", "Simulation"], axis=1
    )
    
    plt.figure(figsize=(12, 9))
    plt.title(type)
    plt.plot(sesdyn_type_df.T)
    plt.legend()
    plt.show()
```

NOTE: GR1D2-3 look weird!

```python
ists_data_dir = os.path.join("BehaviorDatasets", "ISTS_TrainingSet", sample_dir)
ists_type_arr = os.listdir(ists_data_dir)
ists_df = None
for type in ists_type_arr:
    if type.endswith(".txt"):
        ists_type_path = os.path.join(ists_data_dir, type)
        ists_type_df = pd.read_csv(ists_type_path, sep="\t", header=None)
        ists_type_df["behaviour_type"] = type[:-4]

        if ists_df is None:
            ists_df = ists_type_df
        else:
            ists_df = pd.concat((ists_df, ists_type_df), ignore_index=True)
ists_df = ists_df.reindex(columns=["behaviour_type"] + ists_type_df.columns[:-1].tolist())
ists_df
```

```python
for type in ists_df["behaviour_type"].unique():
    ists_type_df = ists_df[ists_df["behaviour_type"]==type]
    ists_type_df = ists_type_df.drop(["behaviour_type"], axis=1)
    
    plt.figure(figsize=(12, 9))
    plt.title(type)
    plt.plot(ists_type_df.T)
    plt.show()
```
